```python
status_percentage = orders["order_status"].value_counts(normalize=True) * 100

print(status_percentage.round(2))
```

```python
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)

orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"]
)
```

```python
orders["delivery_days"] = (
    orders["order_delivered_customer_date"] -
    orders["order_purchase_timestamp"]
).dt.days
```

```python
slow_orders = orders[
    orders["delivery_days"] > 30
]

slow_orders.shape
```

```python
slow_order_count = len(slow_orders)

total_delivered_orders = orders["delivery_days"].count()

slow_order_percentage = (
    slow_order_count / total_delivered_orders
) * 100

print("Slow orders:", slow_order_count)
print("Delivered orders with recorded dates:", total_delivered_orders)
print("Slow order percentage:", round(slow_order_percentage, 2), "%")
```

```python
payments = pd.read_csv("datasets/olist_order_payments_dataset.csv")
```

```python
order_payments = payments.groupby("order_id")["payment_value"].sum().reset_index()
```

```python
revenue_data = orders.merge(
    order_payments,
    on="order_id",
    how="inner"
)
```

```python
revenue_data["purchase_month"] = (
    revenue_data["order_purchase_timestamp"].dt.to_period("M")
)
```

```python
monthly_revenue = (
    revenue_data.groupby("purchase_month")["payment_value"]
    .sum()
)

print(monthly_revenue.head(10))
```

```python
plt.figure(figsize=(12, 5))

monthly_revenue.plot(kind="line", marker="o")

plt.title("Monthly Revenue Trends")
plt.xlabel("Month")
plt.ylabel("Revenue (BRL)")
plt.xticks(rotation=45)
plt.grid(True)

plt.savefig(
    "visualizations/monthly_revenue_trends.png",
    bbox_inches="tight"
)

plt.show()
```
